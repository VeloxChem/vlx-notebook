## Cell types

A Jupyter notebook consists of a number of different cells which can be executed in any order, the most relevant being:

- **Markdown**, where you write text and equations, insert hyperlinks, figures, and more

- **Code**, which contains executable (Python) code

A cell is executed by marking the cell and pressing `shift`+`enter`, or the `run` botton at the top. For a markdown cell this formats the text, and for code cells it runs the code. You can also press `ctrl`+`enter`, which only runs the cell, while `shift`+`enter` runs the cell and then selects the next one. 

Results from cells are saved in memory, so it is often a good idea to separate complicated workflows into several blocks, *e.g.* performing heavy calculations in one block and doing post-processing in another block.

The cell-type is selected in a scroll-down window at the top of the notebook window, which for this cell is `Markdown`. For editing the formatted markdown cells you double-click on the text (or press `enter`), which changes the format to plain text.

## Python functions

Next, let us consider impact of different [basis sets](https://veloxchem.org/docs/basis-sets/) on the total energy, and see if there are any general trends for increasing basis set sizes. First, define a function which takes a molecular structure and basis set, and return the SCF energy and number of basis functions:

````python
def get_scf_energy(mol_string, basis_set):

    molecule = vlx.Molecule.read_molecule_string(mol_string)
    basis = vlx.MolecularBasis.read(molecule, basis_set)

    nbas = basis.get_dimensions_of_basis()

    scf_drv = vlx.ScfRestrictedDriver()
    scf_drv.ostream.mute()

    scf_results = scf_drv.compute(molecule, basis)
    scf_energy = scf_drv.get_scf_energy()

    return scf_energy, nbas
````

## Python lists

- Python lists are collections of objects (numbers, strings, other lists, ...) inside square brackets.
- Lists are mutable and iterable.

Create two empty lists for storing SCF energies and basis set sizes:

````python
scf_energies = []
basis_sizes = []
````

Next, calculate the SCF energies and basis set sizes when using the following basis sets:

- STO-3G   
- 6-31G
- 6-311G
- 6-31++G**
- 6-311++G**
    
Details on what the basis set designations mean can be found on [Wikipedia](https://en.wikipedia.org/wiki/Basis_set_(chemistry)). 

Create a list with the basis set strings and run `get_scf_energy`:

````python
basis_sets = ["STO-3G", "6-31G", "6-311G", "6-31++G**", "6-311++G**"]

for basis_set in basis_sets:

    energy, nbas = get_scf_energy(water_str, basis_set)

    scf_energies.append(energy)
    basis_sizes.append(nbas)
````

## Spline interpolation

````python
scf_drv = vlx.ScfRestrictedDriver()
scf_drv.ostream.mute()
````

An important concept in computational chemistry is that of the [potential energy surface (PES)](https://en.wikipedia.org/wiki/Potential_energy_surface), which describe the energy of a system as a function of some parameter (typically atomic positions). We here wish to construct this for carbon monoxide, by tracking the total energy as a function of the bond length.

````python
# carbon monoxide
molecule = vlx.Molecule.read_smiles("[C-]#[O+]")
basis = vlx.MolecularBasis.read(molecule, "def2-svp")
````

````python
molecule.show(atom_indices=True)
````

````python
energies = []
bond_lengths = np.arange(0.92, 1.40, 0.04)

for bond_length in bond_lengths:
    
    molecule.set_distance_in_angstroms((1, 2), bond_length)
    
    scf_results = scf_drv.compute(molecule, basis)
    energies.append(scf_drv.get_scf_energy())
````

````python
plt.figure(figsize=(5, 3.5))

spline_func = scipy.interpolate.interp1d(bond_lengths, energies, kind="cubic")
x_spline = np.linspace(min(bond_lengths), max(bond_lengths), 100)
y_spline = spline_func(x_spline)

plt.plot(x_spline, y_spline, label="Cubic spline")
plt.plot(bond_lengths, energies, "o", label="QM data")

plt.ylabel("Energy (Hartree)")
plt.xlabel("Bond length (Å)")
plt.legend()
plt.show()
````

## Einstein summation

The eChem book makes use of `np.einsum` as a means to perform [Einstein summation](https://en.wikipedia.org/wiki/Einstein_notation), which is descried in the [manual](https://numpy.org/doc/stable/reference/generated/numpy.einsum.html), and with basic examples found [here](https://ajcr.net/Basic-guide-to-einsum/).

Consider a vector A and a matrix B:

````python
A = np.array([0, 1, 2])
B = np.array([[1, 1, 3], [3, 4, 5], [5, 6, 7]])

print("A:", A)
print("B:\n", B)
````

Say that we want to perform this operation:

$$
C_i = \sum_j A_iB_{ij}
$$

In `np.einsum` this can be performed in a single step:

````python
np.einsum("i, ij -> i", A, B)
````

Here $i$ and $j$ select axis in the objects, and you can change the indices to any letter (provided you use them consistently). The operation carried out takes all $A_i$ and multiply with $B_{ij}$, summing over $j$ and returning a new object with elements along $i$. 

Removing the $i$ on the right-hand side instead gives the total sum:

````python
np.einsum("i, ij ->", A, B)
````

This can be used for simpler operations such as taking the sum of an object, performing element-wise multiplication, calculating the trace, and more. It can also be used for more complicated matrix-matrix operations and considering more arrays, and we will use it to perform a transformation from atomic to molecular orbitals (AO and MO, respectively).

### Slicing NumPy arrays

Perform an SCF calculation for water.

````python
scf_drv = vlx.ScfRestrictedDriver()
scf_drv.ostream.mute()

molecule = vlx.Molecule.read_name("water")
basis = vlx.MolecularBasis.read(molecule, "def2-svp")

scf_results = scf_drv.compute(molecule, basis)
````

Print out the orbital energies.

````python
orbital_energies = scf_results["E_alpha"]
print(orbital_energies)
````

Get the number of occupied orbitals.

````python
nocc = molecule.number_of_alpha_electrons()
````

We [slice](https://www.w3schools.com/python/numpy/numpy_array_slicing.asp) the NumPy array to separate orbital energies for occupied and unoccupied orbitals.

````python
print("Energy of occupied:\n", orbital_energies[:nocc])
print("Energy of unoccupied:\n", orbital_energies[nocc:])
````

### AO to MO transformation

[Integrals](https://kthpanor.github.io/echem/integrals/) are calculated in the atomic orbital (AO) basis, while, in many cases, we prefer to work in the molecular orbital (MO) basis. We then need to perform an AO to MO transformation. 

As an example, consider the expectation value of the kinetic energy:

$$
    \langle \hat{T} \rangle = 
    2 \sum_{i}^\mathrm{occ} T_{ii}
$$

where 

$$
T_{pq} =
\int \phi_p^*(\mathbf{r}) \hat{T} \phi_q(\mathbf{r}) d^3\mathbf{r}
$$

**Note**
> All orbitals are real, so you can ignore the operation of complex conjugatation.

We have access to the kinetic energy in the AO basis:

$$
T_{\alpha\beta}^\mathrm{AO} =
\int \chi_\alpha^*(\mathbf{r}) \hat{T} \chi_\beta(\mathbf{r}) d^3\mathbf{r}
$$

````python
T_ao = vlx.compute_kinetic_energy_integrals(molecule, basis)
````

We calculate the elements of the kinetic energy operator in the MO basis from the transformation:

$$
T_{pq} = \sum_{\alpha,\beta} C^*_{\alpha p} C_{\beta q} \, T^\mathrm{AO}_{\alpha\beta}
$$

````python
C = scf_results["C_alpha"]
````

````python
T_mo = np.einsum("ap, bq, ab -> pq", C, C, T_ao)
````

Finally, the expectation value of the kinetic operator is:

````python
T_exp = 2 * np.einsum("ii ->", T_mo[:nocc, :nocc])

print(f"Expectation value: {T_exp:.6f} Hartree")
````
