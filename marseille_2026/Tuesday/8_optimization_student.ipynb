```python
def get_internal_coordinates(z_matrix):
    """ Creates a list of internal coordinates, based on the z_matrix.

        :param z_matrix: a list of atomic indices which define bonds, angles, and dihedrals.
    """
    internal_coordinates = []
    for z in z_matrix:
        # a bond is defined by 2 atoms
        if len(z) == 2:
            # indexing in geometric starts at 0
            q = geometric.internal.Distance(z[0]-1, z[1]-1)
            internal_coordinates.append(q)
        
        # an bond angle is defined by 3 atoms
        elif len(z) == 3:
            q = geometric.internal.Angle(z[0]-1, z[1]-1, z[2]-1)
            internal_coordinates.append(q)
        
        # a dihedral is defined by 4 atoms
        elif len(z) == 4:
            q = geometric.internal.Dihedral(z[0]-1, z[1]-1, z[2]-1, z[3]-1)
            internal_coordinates.append(q)
        else:
            print("Unrecognized type of coordinate.")
    return internal_coordinates

internal_coordinates = get_internal_coordinates(z_matrix)
print("List of %d non-redundant internal coordinates:\n" % (len(internal_coordinates)))
print(internal_coordinates)
print()
```

To be able to judge the particular bond length, we must transform the gradient form Cartesian to internal coordinates.

```python
def transform_gradient_to_internal(molecule, internal_coordinates, gradient, tol=1e-8):
    """
    Calculates the Wilson B-matrix and converts the gradient Cartesian to internal Coordinates.

    :param molecule: the molecule.
    :param internal_coordinates: the list of internal coordinates (geomeTRIC objects).
    :param gradient: the gradient in Cartesian coordinates.
    """
    n_atoms = molecule.number_of_atoms()
    coords = molecule.get_coordinates_in_angstrom().reshape((n_atoms * 3))

    b_matrix = np.zeros((len(internal_coordinates), n_atoms * 3))

    for i, q in enumerate(internal_coordinates):
        derivative = q.derivative(coords).reshape(-1)
        b_matrix[i] = derivative

    g_matrix = np.dot(b_matrix, b_matrix.T)
    U, s, Vt = np.linalg.svd(g_matrix)

    s_inv = np.array([1 / s_i if s_i > tol else 0.0 for s_i in s])
    
    g_minus_matrix = np.dot(U, np.dot(np.diag(s_inv), Vt))

    gradient_flat = gradient.flatten()
    internal_gradient = np.dot(g_minus_matrix, np.dot(b_matrix, gradient_flat))
    return internal_gradient
```

### <font color=#138D75> 2.3. Molecular geometry optimization </font>
In this section, we will perform an actual geometry optimization. For this, we will need to adapt the Newton-Raphson algorithm to work for a molecule. This requires that we calculate the molecular gradient, which we have done above.

Additionally, the Newton-Raphson routine requires the Hessian matrix as well. Because this is typically quite resource-intensive, especially for larger molecules, the Hessian is actually approximated. A Newton-Raphson algorithm with approximate Hessian is called quasi-Newton.

#### <font color=#138D75> Quasi-Newton </font>
In practical molecular geometry optimizations, the exact Hessian is replaced by an approximate Hessian matrix. This is known as the quasi-Newton algorithm. One popular approximation for the Hessian is the Broyden-Fletcher-Goldfarb-Shanno (BFGS) approximation,
  $$\mathbf{H}^{-1}_{i+1} = \left(\mathbf{I}-\rho \mathbf{s}_i\mathbf{y}^\mathrm{T}_i \right)\mathbf{H}^{-1}_{i}\left(\mathbf{I}-\rho_i \mathbf{y}_i\mathbf{s}^\mathrm{T}_i \right)+\rho_i\mathbf{s}_i\mathbf{s}^\mathrm{T}_i, $$
    where, $$\mathbf{y}_i = \mathbf{g}_{i+1} - \mathbf{g}_i,$$
  $$ \mathbf{s}_i = \mathbf{x}_{i+1} - \mathbf{x}_i,$$
  $$ \rho_i = \mathbf{y}^\mathrm{T}_i\mathbf{s}_i .$$


Once we have the approximate inverse Hessian, we can get the displacement in the same way as in the Newton-Raphson algorithm,
     $$\Delta \mathbf{r}_i = -\mathbf{H}_i^{-1}\mathbf{g}_i,$$
     $$\mathbf{r}_{i+1}=\mathbf{r}_i + \Delta\mathbf{r}_{i}.$$

```python
def bfgs_hessian_inverse(Hi_inv, gi, gi_p1, xi, xi_p1, natm):
    """Implements the Broyden-Fletcher-Goldfarb-Shanno
       approximation for the inverse Hessian matrix.
       
       param Hi_inv  : inverse Hessian at step i.
       param gi      : gradient at step i.
       param gi_p1   : gradient at step i+1.
       param xi      : coordinates at step i.
       param xi_p1   : coordinates at step i+1.
    """
    I = np.identity(3*natm)
    yi = (gi_p1 - gi).reshape(natm*3).reshape(-1, 1) # make a 1D column vector
    si = (xi_p1 - xi).reshape(natm*3).reshape(-1, 1) # make a 1D column vector
    
    rho = 1/np.matmul(yi.T, si)
    
    curvature = np.matmul(si.T, yi)

    I_rho_sy = I - rho * np.matmul(si, yi.T)
    I_rho_ys = I - rho * np.matmul(yi, si.T)
    rho_ss = rho * np.matmul(si, si.T)
    
    Hi_p1_inv = np.linalg.multi_dot([I_rho_sy, Hi_inv, I_rho_ys]) + rho_ss                        
    
    return Hi_p1_inv
```

```python
# Let us prepare some routines which will make it easier to print the values of internal coordinates and compare molecular configurations
def print_comparison(molecule_A, molecule_B, internal_coords, label_A="A", label_B="B"):
    #Compare the initial and optimized geometries
    print("\nInternal coordinate | {:^17s} | {:^17s}".format(label_A, label_B)) 
    print("---------------------------------------------------------------\n")
    for q in internal_coords:
        if isinstance(q, geometric.internal.Distance):
            # Calculate the bond length and transform from au to Angstroms
            print(q, "       |    %8.5f Å     |    %8.5f Å" % ( q.value(molecule_A.get_coordinates_in_angstrom()),
                                         q.value(molecule_B.get_coordinates_in_angstrom())
                                       ))
        if isinstance(q, geometric.internal.Angle):
            # Calculate the bond angle and transform from radians to degrees
            print(q, "        |     %6.1f deg    |     %6.1f deg" % (180/np.pi * q.value(molecule_A.get_coordinates_in_angstrom()),
                                              180/np.pi * q.value(molecule_B.get_coordinates_in_angstrom())
                                             ) )
        if isinstance(q, geometric.internal.Dihedral):
            # Calculate the dihedral and transform from radians to degrees
            print(q, "   |    %7.1f deg    |    %7.1f deg" % (180/np.pi * q.value(molecule_A.get_coordinates_in_angstrom()),
                                              180/np.pi * q.value(molecule_B.get_coordinates_in_angstrom())
                                             ) )
```
